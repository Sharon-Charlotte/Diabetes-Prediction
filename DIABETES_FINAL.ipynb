```python
categorical_features = [
    'sex_binary',
    'age_group',
    'race',
    'education',
    'income_group',
    'employment',
    'marital_status',
    'hypertension',
    'high_cholesterol',
    'general_health',
    'exercise',
    '_SMOKER3'
]
```

```python
numeric_features = [
    'BMI',
    'physical_unhealthy_days',
    'mental_unhealthy_days',
    'fruit_times_day',
    'green_vegetable_times_day',
    'drinks_per_week'
]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining outcome:")
print(y_train.value_counts(normalize=True))

print("\nTesting outcome:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
```

```python
preprocessor = ColumnTransformer([
    ('categorical', categorical_pipeline, categorical_features),
    ('numeric', numeric_pipeline, numeric_features)
])
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])
```

```python
logistic_model.fit(X_train, y_train)

print("Logistic regression model trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Create the Streamlit app file

app_code = r'''
import streamlit as st
import pandas as pd
import joblib

# -----------------------------
# Page configuration
# -----------------------------
st.set_page_config(
    page_title="Diabetes Risk Prediction",
    page_icon="🩺",
    layout="wide"
)

# -----------------------------
# Load model
# -----------------------------
MODEL_PATH = "/content/models/diabetes_logistic_pipeline.joblib"

try:
    model = joblib.load(MODEL_PATH)
except Exception as e:
    st.error(f"Could not load the model: {e}")
    st.stop()

THRESHOLD = 0.20

# -----------------------------
# Title
# -----------------------------
st.title("🩺 Diabetes Risk Prediction Dashboard")
st.write(
    "A logistic regression model developed using the CDC 2021 BRFSS dataset."
)

st.info(
    "The prediction is a model-based estimate and should not be interpreted "
    "as a medical diagnosis."
)

# -----------------------------
# Input form
# -----------------------------
st.header("Enter Individual Information")

col1, col2, col3 = st.columns(3)

with col1:
    sex_binary = st.selectbox(
        "Sex",
        [0, 1],
        format_func=lambda x: "Female" if x == 0 else "Male"
    )

    age_group = st.selectbox(
        "Age Group",
        [1, 2, 3, 4, 5, 6]
    )

    race = st.selectbox(
        "Race Code",
        [1, 2, 3, 4, 5, 6, 7, 8, 9]
    )

    education = st.selectbox(
        "Education Code",
        [1, 2, 3, 4, 5, 6]
    )

    income_group = st.selectbox(
        "Income Group",
        [1, 2, 3, 4, 5, 6, 7]
    )

    employment = st.selectbox(
        "Employment Code",
        [1, 2, 3, 4, 5, 6, 7, 8]
    )

with col2:
    marital_status = st.selectbox(
        "Marital Status Code",
        [1, 2, 3, 4, 5, 6]
    )

    BMI = st.number_input(
        "BMI",
        min_value=10.0,
        max_value=80.0,
        value=25.0,
        step=0.1
    )

    hypertension = st.selectbox(
        "Hypertension",
        [0, 1],
        format_func=lambda x: "No" if x == 0 else "Yes"
    )

    high_cholesterol = st.selectbox(
        "High Cholesterol",
        [0, 1],
        format_func=lambda x: "No" if x == 0 else "Yes"
    )

    general_health = st.selectbox(
        "General Health Code",
        [1, 2, 3, 4, 5]
    )

    physical_unhealthy_days = st.slider(
        "Physical Unhealthy Days",
        0, 30, 0
    )

with col3:
    mental_unhealthy_days = st.slider(
        "Mental Unhealthy Days",
        0, 30, 0
    )

    exercise = st.selectbox(
        "Exercise",
        [0, 1],
        format_func=lambda x: "No" if x == 0 else "Yes"
    )

    fruit_times_day = st.number_input(
        "Fruit Times per Day",
        min_value=0.0,
        max_value=20.0,
        value=1.0,
        step=0.1
    )

    green_vegetable_times_day = st.number_input(
        "Green Vegetable Times per Day",
        min_value=0.0,
        max_value=20.0,
        value=1.0,
        step=0.1
    )

    drinks_per_week = st.number_input(
        "Drinks per Week",
        min_value=0.0,
        max_value=1000.0,
        value=0.0,
        step=1.0
    )

    smoker = st.selectbox(
        "Smoking Status Code",
        [1, 2, 3, 4]
    )

# -----------------------------
# Prediction
# -----------------------------
if st.button("🔍 Predict Diabetes Risk", type="primary"):

    input_data = pd.DataFrame({
        "sex_binary": [sex_binary],
        "age_group": [age_group],
        "race": [race],
        "education": [education],
        "income_group": [income_group],
        "employment": [employment],
        "marital_status": [marital_status],
        "BMI": [BMI],
        "hypertension": [hypertension],
        "high_cholesterol": [high_cholesterol],
        "general_health": [general_health],
        "physical_unhealthy_days": [physical_unhealthy_days],
        "mental_unhealthy_days": [mental_unhealthy_days],
        "exercise": [exercise],
        "fruit_times_day": [fruit_times_day],
        "green_vegetable_times_day": [green_vegetable_times_day],
        "drinks_per_week": [drinks_per_week],
        "_SMOKER3": [smoker]
    })

    probability = model.predict_proba(input_data)[0, 1]

    prediction = int(probability >= THRESHOLD)

    st.header("Prediction Result")

    c1, c2 = st.columns(2)

    with c1:
        st.metric(
            "Predicted Probability",
            f"{probability:.2%}"
        )

    with c2:
        if prediction == 1:
            st.error("Higher-risk classification")
        else:
            st.success("Lower-risk classification")

    st.write(
        f"The model uses a classification threshold of **{THRESHOLD:.2f}**."
    )

# -----------------------------
# Model information
# -----------------------------
st.divider()

st.header("Model Performance")

m1, m2, m3, m4, m5 = st.columns(5)

m1.metric("Accuracy", "79.44%")
m2.metric("Precision", "35.99%")
m3.metric("Recall", "65.86%")
m4.metric("F1 Score", "46.55%")
m5.metric("ROC-AUC", "82.86%")

st.caption(
    "Performance shown at the selected probability threshold of 0.20, "
    "except ROC-AUC which is threshold-independent."
)
'''

with open("/content/app.py", "w") as f:
    f.write(app_code)

print("App created successfully.")
print("App exists:", __import__("os").path.exists("/content/app.py"))
print("App size:", __import__("os").path.getsize("/content/app.py"), "bytes")
```

```python
from pathlib import Path
import shutil

drive = Path("/content/drive/MyDrive")

# Existing source files
notebook_source = drive / "Colab Noteboooks" / "DIABETES_FINAL.ipynb"
model_folder = drive / "Diabetes_Project" / "models"

# Destination project
project = Path("/content/diabetes-prediction-capstone")
notebook_dest = project / "notebooks"
model_dest = project / "models"

notebook_dest.mkdir(parents=True, exist_ok=True)
model_dest.mkdir(parents=True, exist_ok=True)

# Copy original notebook
if notebook_source.exists():
    shutil.copy2(
        notebook_source,
        notebook_dest / "DIABETES_FINAL.ipynb"
    )
    print("Notebook copied successfully.")
else:
    print("Notebook source not found. Check the folder spelling.")

# Copy existing saved model(s)
if model_folder.exists():
    models = [
        p for p in model_folder.iterdir()
        if p.is_file() and p.suffix.lower() in [".joblib", ".pkl"]
    ]

    if models:
        for model in models:
            shutil.copy2(model, model_dest / model.name)
            print("Model copied:", model.name)
    else:
        print("No .joblib or .pkl files found in the model folder.")
else:
    print("Model folder not found.")

print("\nProject folder:", project)
```
